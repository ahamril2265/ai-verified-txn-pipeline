```python
dbutils.widgets.text("batch_date", "2026-01-01")
batch_date = dbutils.widgets.get("batch_date")
```

```python
SIM = "/Volumes/fintxn/ops/sim_batches"
LANDING = "/Volumes/fintxn/bronze/landing"

for feed in ["transactions", "customers"]:
    src = f"{SIM}/{feed}/{feed}_{batch_date}.csv"
    dst = f"{LANDING}/{feed}/{feed}_{batch_date}.csv"
    dbutils.fs.mkdirs(f"{LANDING}/{feed}")
    dbutils.fs.cp(src, dst)
    print("landed", dst)
```
